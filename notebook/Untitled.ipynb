```python
import data_loaders
import matplotlib.pyplot as plt
import pandas as pd
data = data_loaders.DataLoader()
inflow = data.inflow_data.copy()

inflow['date'] = pd.to_datetime(inflow['date'])
inflow_2015 = inflow[inflow['date'].dt.year >= 2015].copy()

# 2. Extract year as a column
inflow_2015['year'] = inflow_2015['date'].dt.year

# 3. List of reservoir columns (excluding date, r_Vest_Vassdraget, and year)
reservoir_cols = [
    'Langesae', 'Staavatn', 'Kjelavatn', 'Bordalsvatn', 'Foersvatn',
    'Venemo', 'Bitdalsvatn', 'Songavatn', 'Totak', 'Vaamarvatn',
    'Langeidvatn', 'Vatjern', 'Vinjevatn', 'Botnedalsvatn', 'Byrtevatn', 'Bandak', 'Hyljelihyl'
]
# Note: 'Hyljelihyl' is excluded as requested earlier. Add it back to reservoir_cols if needed.

# 4. Calculate total inflow per reservoir per year
annual_inflow_by_res = inflow_2015.groupby('year')[reservoir_cols].sum()

# 5. Calculate total inflow across ALL reservoirs combined per year
annual_inflow_by_res['TOTAL_SYSTEM'] = annual_inflow_by_res.sum(axis=1)

print(annual_inflow_by_res)
```
